```python
# --- IMPORTATION DES BIBLIOTHÈQUES ---
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

print("Version TensorFlow :", tf.__version__)

# --- ÉTAPE 1 : Importation de la base de données ---
# Chargement de fashion_mnist à partir de tf.keras.datasets
fashion_mnist = tf.keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# --- ÉTAPE 2 : Afficher la taille des tableaux ---
print("\nTailles initiales :")
print("x_train :", x_train.shape)
print("y_train :", y_train.shape)
print("x_test :", x_test.shape)
print("y_test :", y_test.shape)

# --- ÉTAPE 3 : Normalisation des images ---
# Les valeurs des pixels vont de 0 à 255. On les divise par 255.0 pour les mettre à l'échelle 0 à 1.
x_train = x_train / 255.0
x_test = x_test / 255.0

# --- ÉTAPE 4 : Décomposition en apprentissage (90%) et validation (10%) ---
# On utilise train_test_split pour séparer aléatoirement sans répétition
x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.10, random_state=42)

print("\nTailles après séparation validation :")
print("x_train (90%) :", x_train.shape)
print("y_train (90%) :", y_train.shape)
print("x_val (10%) :", x_val.shape)
print("y_val (10%) :", y_val.shape)
```

```python
# --- ÉTAPE 5 : Afficher les 12 premières images de l'ensemble d'apprentissage ---
# Définition des noms de classes officiels pour Fashion MNIST
class_names = ['T-shirt/top', 'Pantalon', 'Pull', 'Robe', 'Manteau',
               'Sandale', 'Chemise', 'Baskets', 'Sac', 'Bottine']

plt.figure(figsize=(12, 4))
for i in range(12):
    plt.subplot(2, 6, i+1) # 2 lignes, 6 colonnes
    plt.xticks([]) # Masquer les axes
    plt.yticks([])
    plt.grid(False)
    plt.imshow(x_train[i], cmap=plt.cm.binary)
    # Ajouter le nom de la classe en dessous de l'image
    plt.xlabel(class_names[y_train[i]])
plt.tight_layout()
plt.show()

# --- ÉTAPE 6 : Création du modèle cls_img ---
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

cls_img = Sequential([
    # Couche d'entrée : convertit l'image 2D (28x28) en 1D (784)
    Flatten(input_shape=(28, 28)),

    # Première couche cachée (128 neurones)
    Dense(128, activation='relu'),

    # Deuxième couche cachée (128 neurones)
    Dense(128, activation='relu'),

    # Troisième couche cachée (64 neurones)
    Dense(64, activation='relu'),

    # Couche de sortie (10 neurones pour les 10 classes, activation softmax pour les probabilités)
    Dense(10, activation='softmax')
])

# --- ÉTAPE 7 : Résumé de la structure du modèle ---
print("\nRésumé de la structure du modèle :")
cls_img.summary()
```

```python
# --- COMPILATION ET ENTRAÎNEMENT (Prérequis pour l'étape 8) ---
cls_img.compile(optimizer='adam',
                loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])

print("Début de l'entraînement...")
# On entraîne sur 15 époques (vous pouvez ajuster ce nombre)
history = cls_img.fit(x_train, y_train,
                      epochs=15,
                      validation_data=(x_val, y_val))

# --- ÉTAPE 8 : Tracer la courbe d'évaluation (Accuracy et Loss) ---
plt.figure(figsize=(12, 4))

# Courbe de l'Accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Apprentissage')
plt.plot(history.history['val_accuracy'], label='Validation')
plt.title('Évolution de l\'Accuracy')
plt.xlabel('Époque')
plt.ylabel('Accuracy')
plt.legend()

# Courbe de la Perte (Loss)
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Apprentissage')
plt.plot(history.history['val_loss'], label='Validation')
plt.title('Évolution de la Perte (Loss)')
plt.xlabel('Époque')
plt.ylabel('Perte')
plt.legend()

plt.tight_layout()
plt.show()

# --- ÉTAPE 10 : Calculer l'accuracy et la perte (loss) pour les données de test ---
# (Fait avant l'étape 9 dans le code pour réutiliser les prédictions)
print("\n--- Évaluation sur les données de test ---")
test_loss, test_acc = cls_img.evaluate(x_test, y_test, verbose=0)
print(f"Perte (Loss) sur le test : {test_loss:.4f}")
print(f"Accuracy sur le test : {test_acc:.4f}")

# Prédictions sur les données de test pour la matrice et le rapport
y_pred_probs = cls_img.predict(x_test)
y_pred = np.argmax(y_pred_probs, axis=1)

# --- ÉTAPE 9 : Visualiser la matrice de confusion ---
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Matrice de Confusion')
plt.xlabel('Prédictions')
plt.ylabel('Valeurs Réelles')
plt.show()

# --- ÉTAPE 11 : Afficher le rapport de classification ---
print("\n--- Rapport de Classification ---")
print(classification_report(y_test, y_pred, target_names=class_names))
```

```python
from tensorflow.keras.layers import Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping

# --- CRÉATION DU MODÈLE AMÉLIORÉ ---
modele_ameliore = Sequential([
    Flatten(input_shape=(28, 28)),

    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.3), # Désactive 30% des neurones aléatoirement

    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),

    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),

    Dense(10, activation='softmax')
])

# --- COMPILATION ---
modele_ameliore.compile(optimizer='adam',
                        loss='sparse_categorical_crossentropy',
                        metrics=['accuracy'])

# --- EARLY STOPPING ---
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# --- ENTRAÎNEMENT ---
print("Entraînement du modèle amélioré en cours...")
history_amelio = modele_ameliore.fit(x_train, y_train,
                                     epochs=50,
                                     validation_data=(x_val, y_val),
                                     callbacks=[early_stop])

# --- ÉVALUATION ---
print("\n--- Évaluation du modèle amélioré sur les données de test ---")
test_loss_amelio, test_acc_amelio = modele_ameliore.evaluate(x_test, y_test, verbose=0)
print(f"Nouvelle Perte (Loss) : {test_loss_amelio:.4f}")
print(f"Nouvelle Accuracy : {test_acc_amelio:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping

# Noms des classes pour l'affichage
class_names = ['T-shirt/top', 'Pantalon', 'Pull', 'Robe', 'Manteau',
               'Sandale', 'Chemise', 'Baskets', 'Sac', 'Bottine']

# --- 1. CRÉATION DU MODÈLE OPTIMISÉ ---
modele_tp_optimise = Sequential([
    Flatten(input_shape=(28, 28)),

    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),

    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.1),

    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.1),

    Dense(10, activation='softmax')
])

# --- 2. COMPILATION ---
modele_tp_optimise.compile(optimizer='adam',
                           loss='sparse_categorical_crossentropy',
                           metrics=['accuracy'])

# --- 3. EARLY STOPPING ---
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# --- 4. ENTRAÎNEMENT ---
print("Début de l'entraînement du modèle optimisé...\n")
history_optimise = modele_tp_optimise.fit(x_train, y_train,
                                          epochs=30,
                                          validation_data=(x_val, y_val),
                                          callbacks=[early_stop])

# --- 5. ÉVALUATION GLOBALE SUR LE TEST ---
print("\n--- Évaluation finale sur les données de test ---")
test_loss_opt, test_acc_opt = modele_tp_optimise.evaluate(x_test, y_test, verbose=0)
print(f"Perte (Loss) : {test_loss_opt:.4f}")
print(f"Accuracy : {test_acc_opt:.4f}")

# --- 6. TRACÉ DES COURBES (Accuracy et Loss) ---
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history_optimise.history['accuracy'], label='Apprentissage')
plt.plot(history_optimise.history['val_accuracy'], label='Validation')
plt.title('Évolution de l\'Accuracy (Modèle Optimisé)')
plt.xlabel('Époque')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_optimise.history['loss'], label='Apprentissage')
plt.plot(history_optimise.history['val_loss'], label='Validation')
plt.title('Évolution de la Perte (Modèle Optimisé)')
plt.xlabel('Époque')
plt.ylabel('Perte')
plt.legend()

plt.tight_layout()
plt.show()

# --- 7. GÉNÉRATION DES PRÉDICTIONS ---
y_pred_probs_opt = modele_tp_optimise.predict(x_test)
y_pred_opt = np.argmax(y_pred_probs_opt, axis=1)

# --- 8. MATRICE DE CONFUSION ---
cm_opt = confusion_matrix(y_test, y_pred_opt)
plt.figure(figsize=(10, 8))
sns.heatmap(cm_opt, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Matrice de Confusion (Modèle Optimisé)')
plt.xlabel('Prédictions')
plt.ylabel('Valeurs Réelles')
plt.show()

# --- 9. RAPPORT DE CLASSIFICATION ---
print("\n--- Rapport de Classification (Modèle Optimisé) ---")
print(classification_report(y_test, y_pred_opt, target_names=class_names))
```
